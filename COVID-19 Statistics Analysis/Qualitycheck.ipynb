# 5-DATA VALIDATION CHECKLIST

### **5.1-CHECKS FOR NEGATIVE VALUES** ###

```python
print((covid["Confirmed"] < 0).sum())
print((covid["Cured"] < 0).sum())
print((covid["Deaths"] < 0).sum())
```

```python
testing.select_dtypes(include="number").lt(0).sum()
```

```python
vaccine.select_dtypes(include="number").lt(0).sum()
```

### **5.2-CHECK LOGICAL ERRORS** ###

```python
covid[
    (covid["Cured"] > covid["Confirmed"]) |
    (covid["Deaths"] > covid["Confirmed"])
]
```

```python
vaccine[vaccine["Second Dose Administered"] > vaccine["First Dose Administered"]]
```

### **5.3-VALIDATE DATE** ###

```python
covid["Date"] = pd.to_datetime(covid["Date"])

testing["Date"] = pd.to_datetime(testing["Date"])

vaccine["Updated On"] = pd.to_datetime(vaccine["Updated On"])
```

# 6-CLEANING DATASET

```python
covid_clean = covid.drop_duplicates().dropna()
```

```python
testing_clean = testing.drop_duplicates().dropna()  
```

```python
vaccine_clean = vaccine.drop_duplicates().dropna()
```

```python
covid_clean.to_csv("covid_clean.csv", index=False)

testing_clean.to_csv("testing_clean.csv", index=False)

vaccine_clean.to_csv("vaccine_clean.csv", index=False)
```
